**เป้าหมาย:** ทำความรู้จักโครงสร้างของชุดข้อมูลก่อนนำไปวิเคราะห์เชิงลึกต่อไป

ชุดข้อมูลนี้ประกอบด้วยข้อมูลของนักเรียน 50,000 คน จำนวน 16 คุณลักษณะ ครอบคลุมด้านผลการเรียน
พฤติกรรมการใช้ Generative AI พฤติกรรมการเรียน บริบทของสถาบัน และสุขภาพจิต/ความเป็นอยู่ที่ดี


```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

### 1. โหลดชุดข้อมูล (Load Dataset)

```python
df = pd.read_csv("dataset.csv")
df.head()
```

### 2. แสดงขนาดของข้อมูล (Display Shape)

```python
print("จำนวนแถว (นักเรียน) และคอลัมน์ (คุณลักษณะ):", df.shape)
```

### 3. แสดงชนิดข้อมูล (Display Data Types)

```python
df.dtypes
```

### 4. แสดงสถิติเชิงสรุป (Display Summary Statistics)

```python
df.describe()
```

```python
df.describe(include="object")
```

### 5. แสดงค่าที่ขาดหายไป (Display Missing Values)

```python
df.isnull().sum()
```

### 6. แสดงข้อมูลที่ซ้ำกัน (Display Duplicate Records)

```python
print("จำนวนแถวที่ซ้ำกัน:", df.duplicated().sum())
```

### 7. แสดงการกระจายตัวของคลาส (Display Class Distribution)

```python
categorical_cols = [
    "Major_Category", "Year_of_Study", "Primary_Use_Case",
    "Prompt_Engineering_Skill", "Institutional_Policy", "Burnout_Risk_Level"
]

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
```

- คอลัมน์หมวดหมู่ (สาขาวิชา, ชั้นปี, นโยบายสถาบัน, ระดับความเสี่ยง Burnout) มีการกระจายตัวที่หลากหลาย
